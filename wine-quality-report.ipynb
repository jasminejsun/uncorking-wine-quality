## Dataset, problem, and performance metrics

### Dataset chosen

For this project, we will be looking at the UCI Wine Quality dataset (https://archive.ics.uci.edu/dataset/186/wine+quality). The dataset contains red and white Vinho Verde wine samples and their respective physicochemical traits.

Features include:

- Fixed acidity
- Volatile acidity
- Citric acid
- Residual sugar
- Chlorides
- Free sulfur dioxide
- Total sulfur dioxide
- Density
- pH
- Sulphates
- Alcohol

The target variable is a quality variable, which is a score between 0 and 10.

### Why the dataset is challenging

First, wines with different quality scores can share very similar chemical profiles, so perfect separation is difficult/not possible. As well, many features are strongly correlated (e.g., density and alcohol), which motivates dimensionality reduction.

### Problem definition

Which chemical properties best predict wine quality (density, acidity, alcohol)? We want to predict whether a wine is “high quality” (label 1) or “lower quality” (label 0) from its chemical properties.

### Performance metric

Our primary performance metric is classification accuracy on the test set.

## Results

The Random Forest baseline achieved 88.16% test accuracy on the white wine dataset, consistent with the idea that multiple chemical properties jointly influence wine quality. Feature importance shows that alcohol and density are the strongest predictors.

PCA revealed that the dataset has moderate intrinsic dimensionality. The cumulative explained variance plateaus around eight components. This means the data cannot be compressed into only 2–3 meaningful dimensions, but does not require all eleven features either. This reflects that wine chemistry is governed by several partially independent processes (acidity, sulphites, sugar, alcohol, density).

The PCA-based 1-NN accuracy curve mirrors this. Accuracy improves gradually, but stabilizes around 7–8 components, reinforcing that the predictive signal is spread across many dimensions rather than concentrated in a small subspace.

Non-linear dimensionality reduction (UMAP) showed that class structure remains difficult to separate visually. Both 2D and 3D UMAP embeddings show heavy overlap between quality classes, and the accuracy curve plateaus around ~81%.

Overall, the original, linear, and non-linear results consistently show that wine quality does not form clear geometric clusters and that predictive information is moderately high-dimensional, distributed across many chemical attributes.

## Generalization Using External Data (Red Wine)

To test generalization, the PCA+1-NN classifier trained on red wine samples was evaluated using the same methodology used for white wines. Accuracy for red wine increases much more rapidly with dimensionality than in the white-wine experiments. High accuracy is reached within the first 3–4 PCA components, compared to ~7–8 components required for white wine.

This suggests that red-wine quality may have lower intrinsic dimensionality, or that its chemical properties correlate more tightly with perceived quality. Importantly, although the absolute accuracy values differ between the two wine types, both experiments show that predictive information is present, but wine quality cannot be reduced to a small handful of physicochemical properties.

This external test provides evidence that the modeling approach generalizes reasonably well across related datasets.

## What Prevents Perfect Results?

- High-quality wines make up less of the dataset compared to low-quality wines. This places limitations on 1-NN since it predicts labels by looking at the single closest training sample, the majority class tends to dominate these local neighborhoods.
- Even strong predictors like alcohol and density do not perfectly distinguish good from bad wines.
- Since wine quality spreads across many weakly correlated chemical dimensions, the data cannot be represented in a low-dimensional space, which inherently prevents perfect classification.

## Additional Observations

- The Random Forest baseline performs best because it can combine many small signals across features, instead of relying on geometric separation.
- Both PCA and UMAP show substantial overlap between high- and low-quality samples. Even non-linear methods cannot form clean clusters, indicating the difficulty is inherent to the data.
